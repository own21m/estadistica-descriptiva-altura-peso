```
# Calcula el rango de las variables numéricas.
print("El rango de las variables numéricas es:")
print(f'Height(Inches): {df["Height(Inches)"].max() - df["Height(Inches)"].min()}')
print(f'Weight(Pounds): {df["Weight(Pounds)"].max() - df["Weight(Pounds)"].min()}')
```

```
# Crear una nueva columna llamada 'IMC_CLASS' que clasifique los datos según altura y peso en las siguientes categorías: Bajo peso, Normal, Sobrepeso y Obesidad.
imc = df['Weight(Pounds)'] / (df['Height(Inches)'] * df['Height(Inches)']) * 703
df['IMC_CLASS'] = pd.cut(imc, bins=[0, 18.5, 24.9, 29.9, float('inf')], labels=['Bajo peso', 'Normal', 'Sobrepeso', 'Obesidad'])
df.head()
```

18. Calcular el rango intercuartílico de las columnas numéricas.


```
# Calcular el primer cuartil (Q1) y el tercer cuartil (Q3) de la columna Height(Inches) y Weight(Pounds)
Q1_height = df['Height(Inches)'].quantile(0.25)
Q3_height = df['Height(Inches)'].quantile(0.75)

Q1_weight = df['Weight(Pounds)'].quantile(0.25)
Q3_weight = df['Weight(Pounds)'].quantile(0.75)

# Calcular los valores intercuartilicos de las columnas Height(Inches) y Weight(Pounds)
print(f'Rango intercuartilico de Height(Inches): {Q3_height-Q1_height}')
print(f'Rango intercuartilico de Weight(Pounds): {Q3_weight-Q1_weight}')
```

```
# Crear dos nuevas columnas llamadas Outliers_Height y Outliers_Weight:
df['Outliers_Height'] = 0
df['Outliers_Weight'] = 0
df
```

```
# • Outliers_Height se debe calcular usando el método de valores estandarizados sobre la columna Height.
# Calcular el Z-score para la columna Height(Inches)
Z_Score_Height = (df['Height(Inches)'] - df['Height(Inches)'].mean()) / df['Height(Inches)'].std()

# Identificar outliers en Height(Inches) usando el Z-score (marcar como 1 si el valor absoluto del Z-score es mayor que 3)
df['Outliers_Height'] = Z_Score_Height.apply(lambda x: 1 if abs(x) > 3 else 0)

# Mostrar las primeras filas para verificar los resultados
df
```

```
# Calcular Q1, Q3 y el IQR para Weight(Pounds)
Q1_weight = df['Weight(Pounds)'].quantile(0.25)
# Calcula el primer cuartil (Q1) para la columna 'Weight(Pounds)', que corresponde al 25% de los datos.

Q3_weight = df['Weight(Pounds)'].quantile(0.75)
# Calcula el tercer cuartil (Q3) para la columna 'Weight(Pounds)', que corresponde al 75% de los datos.

IQR_weight = Q3_weight - Q1_weight
# Calcula el rango intercuartil (IQR) restando Q1 de Q3. El IQR mide la dispersión central de los datos.

lower_limit = Q1_weight - 1.5 * IQR_weight
# Calcula el límite inferior para identificar valores atípicos, restando 1.5 veces el IQR a Q1.

upper_limit = Q3_weight + 1.5 * IQR_weight
# Calcula el límite superior para identificar valores atípicos, sumando 1.5 veces el IQR a Q3.

df['Outliers_Weight'] = df['Weight(Pounds)'].apply(lambda x: 1 if x < lower_limit or x > upper_limit else 0)
# Crea o actualiza la columna 'Outliers_Weight' en el DataFrame, marcando con 1 los valores de 'Weight(Pounds)' que están fuera de los límites definidos como valores atípicos, y con 0 aquellos que están dentro.

df.head()
```

```
#  Agrupar el conjunto de datos por la columna IMC y usar el método de agregación conteo.
df_grouped = df.groupby('IMC_CLASS').count()
# Agrupa el DataFrame 'df' por la columna 'IMC_CLASS' y cuenta el número de ocurrencias en cada grupo. Los resultados se almacenan en 'df_grouped', mostrando el número de entradas para cada categoría de IMC.

df_grouped
```

```
# ¿Qué porcentaje de personas tiene un peso normal?
peso_normal = df[df['IMC_CLASS'] == 'Normal'].shape[0]
# Filtra el DataFrame 'df' para incluir solo las filas donde la columna 'IMC_CLASS' es 'Normal', y cuenta el número de estas filas usando 'shape[0]'. El resultado se almacena en la variable 'peso_normal'.

print(f'porcentaje de peso normal es de {(peso_normal / df.shape[0]) * 100}%')
# Calcula y muestra el porcentaje de personas clasificadas con 'Normal' en la columna 'IMC_CLASS', dividiendo el número de personas con peso normal entre el total de filas en 'df', y multiplicando por 100 para obtener el porcentaje.
```
