```python
# Calcular Q1, Q3 y el IQR para Weight(Pounds)
Q1_weight = df['Weight(Pounds)'].quantile(0.25)
# Calcula el primer cuartil (Q1) para la columna 'Weight(Pounds)', que corresponde al 25% de los datos.

Q3_weight = df['Weight(Pounds)'].quantile(0.75)
# Calcula el tercer cuartil (Q3) para la columna 'Weight(Pounds)', que corresponde al 75% de los datos.

IQR_weight = Q3_weight - Q1_weight
# Calcula el rango intercuartil (IQR) restando Q1 de Q3. El IQR mide la dispersión central de los datos.

lower_limit = Q1_weight - 1.5 * IQR_weight
# Calcula el límite inferior para identificar valores atípicos, restando 1.5 veces el IQR a Q1.

upper_limit = Q3_weight + 1.5 * IQR_weight
# Calcula el límite superior para identificar valores atípicos, sumando 1.5 veces el IQR a Q3.

df['Outliers_Weight'] = df['Weight(Pounds)'].apply(lambda x: 1 if x < lower_limit or x > upper_limit else 0)
# Crea o actualiza la columna 'Outliers_Weight' en el DataFrame, marcando con 1 los valores de 'Weight(Pounds)' que están fuera de los límites definidos como valores atípicos, y con 0 aquellos que están dentro.

df.head()
```
